```python
import matplotlib.pyplot as plt
import numpy as np
bar_width=0.4
men = [22,30,35,35,26]
women = [25,32,30,35,29]
groups = ['G1','G2','G3','G4','G5']
x = np.arange(len(groups))
plt.bar(x-bar_width/2,men,bar_width,label="Men")
plt.bar(x+bar_width/2,women,bar_width,label="Women")
plt.xticks(x,groups)
plt.ylabel("Scores")
plt.legend()
plt.title("Gender groups")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(0,10,100)
plt.plot(x,np.sin(x),label='sin(x)')
plt.plot(x,np.cos(x),label='cos(x)')
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.title("Q4 Multi Lines")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Two Lines")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

language = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

plt.pie(popularity, labels=language, autopct="%1.1f%%")
plt.title("Programming Language Popularity")
plt.show()
```

```python
import matplotlib.pyplot as plt

language = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

x = range(len(language))

plt.scatter(x, popularity)
plt.xticks(x, language)
plt.xlabel("Language")
plt.ylabel("Popularity")
plt.title("Programming Language Popularity")
plt.show()
```
