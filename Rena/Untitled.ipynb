```python
a=int(input ("Enter a number:"))
b=int(input("Enter your number:"))
ch=int(input("Enter your choice\n1.add\n2.Substract\n3.Multiply\n4.Divide:"))
if ch == 1:
    print("Sum is:", a+b)
elif ch == 2:
    print("sub is:", a-b)
elif ch == 3:
    print("product is:", a*b)
elif ch == 4:
    print("Division is:", a/b)
else:
    print("Invalid choice")
```
