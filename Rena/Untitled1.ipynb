```python
import numpy as np
a = np.array([[1, 2],[4, 3]])
U, S, VT = np.linalg.svd(a)
print("U =\n", U)
print("Singular Values =", S)
print("VT =\n", VT)
import numpy as np

A = np.array([[1, 2],
              [4, 3]])

U, S, VT = np.linalg.svd(A)

Sigma = np.diag(S)

A_reconstructed = U @ Sigma @ VT

print("Original Matrix:\n", A)
print("\nReconstructed Matrix:\n", A_reconstructed)
```
